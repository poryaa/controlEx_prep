```python
from sklearn.datasets import load_sample_images
import matplotlib.pyplot as plt

dataset = load_sample_images()
images = dataset.images

# Show all images in a row
plt.figure(figsize=(8, 4))
for i, img in enumerate(images):
    plt.subplot(1, len(images), i + 1)
    plt.imshow(img)
    plt.axis("off")
plt.tight_layout()
plt.show()  # displays the sample images [web:1][web:13]
```

```python
import tensorflow as tf
images = tf.keras.layers.CenterCrop(height=70, width=120)(images)
images.shape
```

```python
conv_layer = tf.keras.layers.Conv2D(filters=32, kernel_size=(3,3))
fmaps = conv_layer(images)
fmaps.shape
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(200, 300, 3)),
    tf.keras.layers.Conv2D(filters=100, kernel_size=(3,3), strides=2, padding="same"),
    tf.keras.layers.Conv2D(filters=200, kernel_size=(3,3), strides=2, padding="same"),
    tf.keras.layers.Conv2D(filters=300, kernel_size=(3,3), strides=2, padding="same")
])

model.summary()
```
